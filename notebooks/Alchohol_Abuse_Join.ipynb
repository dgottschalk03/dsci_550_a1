```python
import pandas as pd

# Get the 2 datasets (alchohol abuse and haunted places) 
alcohol_abuse_path = "../data/joined_datasets/alchohol_abuse_dataset.csv"
haunted_places_path = "../data/processed/haunted_places_cleaned.tab"

# Load datasets
alcohol_abuse_df = pd.read_csv(alcohol_abuse_path)
haunted_places_df = pd.read_csv(haunted_places_path, sep="\t")  # Tab-separated file

# Standardize column names for merging
alcohol_abuse_df.rename(columns={"State": "state"}, inplace=True)

# Merge datasets on 'state' column
merged_df = pd.merge(haunted_places_df, alcohol_abuse_df, on="state", how="inner")

# Save dataset
merged_df.to_csv("../data/processed/merged_haunted_alcohol.csv", index=False)

# Display the first few rows
print(merged_df)
```
