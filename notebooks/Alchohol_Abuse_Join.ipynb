```python
import pandas as pd

## Copy this code into your preamble. 

# Output Df
outfile = "../data/processed/haunted_places_features_added.tab"

# Reading CSV
haunted_places_df = pd.read_csv("../data/processed/haunted_places_cleaned.tab", sep = "\t")

# Feature Names
feature_names = ["Total Deaths", "% Under 21"]


# Get the 2 datasets (alchohol abuse and haunted places) 
alcohol_abuse_path = "../data/joined_datasets/alchohol_abuse_dataset.csv"
#haunted_places_path = "../data/processed/haunted_places_cleaned.tab"

# Load datasets
alcohol_abuse_df = pd.read_csv(alcohol_abuse_path)
#haunted_places_df = pd.read_csv(haunted_places_path, sep="\t")  # Tab-separated file

# Standardize column names for merging
alcohol_abuse_df.rename(columns={"State": "state"}, inplace=True)

# Merge datasets on 'state' column
merged_df = pd.merge(haunted_places_df, alcohol_abuse_df, on="state", how="inner")

# Save dataset
#merged_df.to_csv("../data/processed/merged_haunted_alcohol.csv", index=False)

## Save CSV ##
print("Saving to CSV...")

# Read Feature added dataframe
out_df = pd.read_csv(f"{outfile}", sep = "\t")


# Check if feature exists
for feature in feature_names:
    
    # If it exists, update values
    if feature in out_df.columns:
        out_df[feature].update(haunted_places_df[feature].values)

    # If not add entire column
    else:
        out_df[feature] = merged_df.loc[:,feature]

out_df.to_csv(f"{outfile}", sep = "\t", index = False)

print(f"CSV Saved to {outfile}")

# Display the first few rows
print(merged_df)
```
