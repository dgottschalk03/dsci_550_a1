```python
import json
import pandas as pd


# Load JSON file
with open("../clustering/jaccard/visualization/circle.json", "r") as f:
    data = json.load(f)

# Function to extract haunted_place_Id values
def extract_haunted_place_ids(data):
    ids = []
    if isinstance(data, dict):
        if "Haunted_Places_Id" in data:
            ids.append(data["Haunted_Places_ID"])
        if "children" in data:
            for child in data.get("children", []):
                ids.extend(extract_haunted_place_ids(child))
    return ids

# Extract IDs and convert to DataFrame
haunted_place_ids = extract_haunted_place_ids(data)
df = pd.DataFrame(haunted_place_ids, columns=["haunted_place_Id"])

# Display the DataFrame
print(df)
```

```python
with open("../clustering/jaccard/visualization/clusters.json", "r") as f:
    data = json.load(f)

clusters = {}

for cl in data.get("children", []):
    clusters[cl['name']] = []
    for entry in cl.get("children", []):
        with open("." + entry['name'].split()[0], 'r') as haunted_place:
            idx = int(json.load(haunted_place)["Haunted_Places_Id"])
            clusters[cl['name']].append(idx)
print(clusters)
```
