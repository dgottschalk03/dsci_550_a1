1.05: Apparition Type

**Apparition Type:**
*   Used keywords specified in apparition_type.txt
*   Search for the following words: ghost, spirit, phantom, male ghost, man ghost, male apparition, female ghost, woman ghost, lady spirit, child ghost, boy spirit, girl ghost, young apparition, orb, floating light, ufo, unidentified flying object, uap, demon, evil presence, dark entity





```python
import pandas as pd
import re
import time
# Output Df
outfile = "../data/processed/haunted_places_features_added.tab"

# Reading CSV
df = pd.read_csv("../data/processed/haunted_places_cleaned.tab", sep = "\t")

# Feature Names
feature_names = ["Apparition_Type"]


# Define apparition keyword patterns with varying tenses
apparition_patterns = {
    "Ghost": r"ghost(?:s)?",
    "Spirit": r"spirit(?:s)?",
    "Phantom": r"phantom(?:s)?",
    "Male Ghost": r"(?:male|man)\s+ghost(?:s)?",
    "Male Apparition": r"male\s+apparition(?:s)?",
    "Female Ghost": r"(?:female|woman)\s+ghost(?:s)?",
    "Lady Spirit": r"lady\s+spirit(?:s)?",
    "Child Ghost": r"child\s+ghost(?:s)?",
    "Boy Spirit": r"boy\s+spirit(?:s)?",
    "Girl Ghost": r"girl\s+ghost(?:s)?",
    "Young Apparition": r"young\s+apparition(?:s)?",
    "Orb": r"orb(?:s)?",
    "Floating Light": r"floating\s+light(?:s)?",
    "UFO": r"ufo(?:s)?",
    "Unidentified Flying Object": r"unidentified\s+flying\s+object(?:s)?",
    "UAP": r"uap(?:s)?",
    "Demon": r"demon(?:s)?",
    "Evil Presence": r"evil\s+presence(?:s)?",
    "Dark Entity": r"dark\s+entity(?:s)?"
}

def classify_apparition(text):
    if not isinstance(text, str):
        return "Unknown"

    text = text.lower()
    matched_types = set()

    # Check every keyword pattern in description
    for label, pattern in apparition_patterns.items():
        # Use word boundaries
        if re.search(rf"\b{pattern}\b", text):
            matched_types.add(label)

    if not matched_types:
        return "Unknown"

    # Return all matched types as a comma-separated string
    return ", ".join(sorted(matched_types))

# Start timing
start = time.time()

# Apply classification to  "description" column.
df["Apparition_Type"] = df["Description"].apply(classify_apparition)

# Stop timing
end = time.time()

# Count occurrences
counts = df["Apparition_Type"].value_counts()
unknown_counts = counts.get("Unknown", 0)

extract_printout = [(category, count) for category, count in counts.items()]
print("-" * 100, "Extraction Completed", "-" * 100)
print(f"Extraction Took: {end - start:.6f} seconds\n")
print("\n".join([f"{category}: {count}" for category, count in extract_printout]))
print("-" * 100)
```

```python
## Save CSV ##
print("Saving to CSV...")

# Read Feature added dataframe
out_df = pd.read_csv(f"{outfile}", sep = "\t")


# Check if feature exists
for feature in feature_names:

    # If it exists, update values
    if feature in out_df.columns:
        out_df[feature].update(df[feature].values)
        out_df[feature] = df[feature].values

    # If not add entire column
    else:
        out_df[feature] = df[feature].values

out_df.to_csv(f"{outfile}", sep = "\t", index = False)

print(f"CSV Saved to {outfile}")
```
