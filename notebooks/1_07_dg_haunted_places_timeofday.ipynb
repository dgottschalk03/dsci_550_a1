1.07 - Time of Day
We defined the following keyword patterns


*   Morning: morning, dawn, sunrise
*   Evening: evening, night, midnight, late
*   Dusk: dusk, sunset, twilight
Otherwise: the time of day is set to unknown






```python
import pandas as pd
import re
import time

# Output Df
outfile = "../data/processed/haunted_places_features_added.tab"

# Reading CSV
df = pd.read_csv("../data/processed/haunted_places_cleaned.tab", sep = "\t")

# Feature Names
feature_names = ["Time of Day"]

# Define  keyword patterns
time_patterns = {
    "Morning": r"\bmorning\b|\bdawn\b|\bsunrise\b",
    "Evening": r"\bevening\b|\bnight\b|\bmidnight\b|\blate\b",
    "Dusk": r"\bdusk\b|\bsunset\b|\btwilight\b"
}

def classify_time_of_day(text):
    if not isinstance(text, str):
        return "Unknown"

    text = text.lower()

    for label, pattern in time_patterns.items():
        if re.search(pattern, text):
            return label  # Return first match found

    return "Unknown"

# Start timing
start = time.time()

# Apply classification to description column
df["Time of Day"] = df["description"].apply(classify_time_of_day)

# Stop timing
end = time.time()

# Count occurrences
counts = df["Time of Day"].value_counts()
unknown_counts = counts.get("Unknown", 0)

extract_printout = [(category, count) for category, count in counts.items()]
print("-" * 100, "Extraction Completed", "-" * 100)
print(f"Extraction Took: {end - start:.6f} seconds\n")
print("\n".join([f"{category}: {count}" for category, count in extract_printout]))
print("-" * 100)
```

Report and Save

```python
## Save CSV ##
print("Saving to CSV...")

# Read Feature added dataframe
out_df = pd.read_csv(f"{outfile}", sep = "\t")


# Check if feature exists
for feature in feature_names:

    # If it exists, update values
    if feature in out_df.columns:
        out_df[feature].update(df[feature].values)
        out_df[feature] = df[feature].values

    # If not add entire column
    else:
        out_df[feature] = df[feature].values

out_df.to_csv(f"{outfile}", sep = "\t", index = False)

print(f"CSV Saved to {outfile}")
```
