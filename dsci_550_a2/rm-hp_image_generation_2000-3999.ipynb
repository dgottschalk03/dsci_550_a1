Generate haunted place images for IDs 2000–3999
 - Uses the Replicate API to generate AI images with Stable Diffusion for each haunted place.
 - Constructs custom prompts based on the description and apparition type of each location.
 - Iterates through dataset rows 2000 to 3999 to generate 2000 images.
 - Saves each image to the `data/generated_images` folder with a unique filename using the Haunted Place ID. 


```python
#generate images for IDs 2000-3999 w/ Stable Diffusion 

# Required packages
import replicate
import requests
import os
import pandas as pd

# set API token 
os.environ["REPLICATE_API_TOKEN"] = "put api key here"

# Load dataset 
df = pd.read_csv("data/processed/haunted_places_features_added_v2.tab", sep="\t")

# Output directory
output_dir = "data/generated_images"
os.makedirs(output_dir, exist_ok=True)

# make the prompt 
def make_prompt(desc, apparition_type):
    if pd.notna(apparition_type) and apparition_type.strip():
        return f"Haunted place illustration featuring {apparition_type.strip()}: {desc}"
    else:
        return f"Haunted place illustration: {desc}"

# function for generating the images 
def generate_sd_image(prompt, image_path):
    try:
        print(f"🌀 Generating image...\nPrompt: {prompt[:100]}...\n")
        output = replicate.run(
            "stability-ai/stable-diffusion:ac732df83cea7fff18b8472768c88ad041fa750ff7682a21affe81863cbe77e4",
            input={
                "width": 768,
                "height": 768,
                "prompt": prompt,
                "scheduler": "K_EULER",
                "num_outputs": 1,
                "guidance_scale": 7.5,
                "num_inference_steps": 50
            }
        )

        if isinstance(output, list) and len(output) > 0:
            image_url = output[0]
            img_data = requests.get(image_url).content
            with open(image_path, 'wb') as f:
                f.write(img_data)
            print(f"Done: Image saved to {image_path}\n")
        else:
            print(" No image returned by Replicate.\n")

    except Exception as e:
        print(f"Failed to generate image: {e}\n")

# Loop through rows 2000–3999
for idx in range(2000, 4000):
    try:
        row = df.loc[idx]
        haunted_id = str(row["Haunted_Places_Id"])
        prompt = make_prompt(row["Description"], row["Apparition_Type"])
        filename = f"hpimg_{haunted_id}.png"
        image_path = os.path.join(output_dir, filename)

        # skip if image exists already 
        if os.path.exists(image_path):
            print(f" Skipping {filename}, already exists.")
            continue

        generate_sd_image(prompt, image_path)

    except Exception as e:
        print(f" Skipping row {idx}: {e}\n")
```
